```python
import cv2
import os
import shutil

# Original dataset path
dataset_path = "/content/Food_Classification_dataset_01"

# New dataset path
new_dataset_path = "/content/New_Food_Dataset_01"

# create new dataset folder
os.makedirs(new_dataset_path, exist_ok=True)

list_1 = []
list_2 = []

for folder in os.listdir(dataset_path):

    food_name = os.path.join(dataset_path, folder)

    # check folder or not
    if os.path.isdir(food_name):

        # create same folder in new dataset
        new_folder_path = os.path.join(new_dataset_path, folder)
        os.makedirs(new_folder_path, exist_ok=True)

        # get image files only
        images = [
            img for img in os.listdir(food_name)
            if img.lower().endswith(('.jpg', '.jpeg', '.png'))
        ]

        # keep only 300 images
        images = images[:300]

        # copy images to new dataset
        for img in images:

            old_img_path = os.path.join(food_name, img)
            new_img_path = os.path.join(new_folder_path, img)

            shutil.copy(old_img_path, new_img_path)

        # store folder names and image counts
        list_1.append(folder)
        list_2.append(len(images))

        print(f"{folder} --> {len(images)} images copied")


print("\nNew Dataset Created Successfully")
```

```python
import os
import shutil
import random

# =========================
# ORIGINAL DATASET PATH
# =========================
dataset_path = "/content/New_Food_Dataset_01"

output_path = "/content/dataset_split_01"

train_path = os.path.join(output_path, "train")
val_path = os.path.join(output_path, "val")


os.makedirs(train_path, exist_ok=True)
os.makedirs(val_path, exist_ok=True)



print("\nSplitting Dataset...")

for folder in os.listdir(new_dataset_path):

    folder_path = os.path.join(new_dataset_path, folder)

    if os.path.isdir(folder_path):

        # create class folders
        os.makedirs(os.path.join(train_path, folder), exist_ok=True)
        os.makedirs(os.path.join(val_path, folder), exist_ok=True)

        # get image files
        images = [
            img for img in os.listdir(folder_path)
            if img.lower().endswith(('.jpg', '.jpeg', '.png'))
        ]

        # shuffle images
        random.shuffle(images)

        # split index
        split_index = int(0.8 * len(images))

        # split images
        train_images = images[:split_index]
        val_images = images[split_index:]

        # copy train images
        for img in train_images:

            src = os.path.join(folder_path, img)
            dst = os.path.join(train_path, folder, img)

            shutil.copy(src, dst)

        # copy validation images
        for img in val_images:

            src = os.path.join(folder_path, img)
            dst = os.path.join(val_path, folder, img)

            shutil.copy(src, dst)

        print(f"{folder}")
        print(f"Train Images      : {len(train_images)}")
        print(f"Validation Images : {len(val_images)}")
        print("-" * 35)

print("\nDataset Split Completed Successfully!")
```

```python
# u should not give the image directly to architeture first give to ImageDataGenerator
# By ImageDataGenerator -> for image we apply some filps, rotations, scale down the pixel values between 0 and 1

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator

```

```python
# Define the Rules
training_data_preprocessing = ImageDataGenerator(1/255,
                                   rotation_range=0.2,
                                   shear_range=0.2,
                                   horizontal_flip=True)
validation_data_preprocessing = ImageDataGenerator(1/255)
```

```python
classes = [
    folder for folder in os.listdir(training_data_path)
    if os.path.isdir(os.path.join(training_data_path, folder))
]

print(classes)

labels = classes
```

```python
# Applying the rules on training data and validation data

final_training_data = training_data_preprocessing.flow_from_directory(training_data_path,
                                                target_size=(200,200),
                                                classes=labels,
                                                class_mode='categorical',
                                                batch_size=32)


final_valdiation_data = validation_data_preprocessing.flow_from_directory(val_data_path,
                                                target_size=(200,200),
                                                classes=labels,
                                                class_mode='categorical',
                                                batch_size=32)
```

```python
# Architecture of Binary CNN

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, Dense, MaxPool2D, Flatten, Activation
from tensorflow.keras.activations import relu, sigmoid, softmax
```

```python
len(labels)
```

```python
model = Sequential()

# Hidden Layer - 1 (applying filter and max pooling)
model.add(Conv2D(filters=500,
                 kernel_size=(3,3),
                 strides=(1,1),
                 padding='valid',
                 kernel_initializer='he_uniform',
                 activation=relu))
model.add(MaxPool2D(pool_size=(2, 2)))

# Hidden Layer - 2
model.add(Conv2D(filters=250,
                 kernel_size=(3,3),
                 strides=(1,1),
                 padding='valid',
                 kernel_initializer='he_uniform',
                 activation=relu))
model.add(MaxPool2D(pool_size=(2, 2)))

# Hidden Layer - 3
model.add(Conv2D(filters=100,
                 kernel_size=(3,3),
                 strides=(1,1),
                 padding='valid',
                 kernel_initializer='he_uniform',
                 activation=relu))
model.add(MaxPool2D(pool_size=(2, 2)))

# Hidden Layer - 4
model.add(Conv2D(filters=50,
                 kernel_size=(3,3),
                 strides=(1,1),
                 padding='valid',
                 kernel_initializer='he_uniform',
                 activation=relu))
model.add(MaxPool2D(pool_size=(2, 2)))

# Hidden Layer - 5
model.add(Conv2D(filters=20,
                 kernel_size=(3,3),
                 strides=(1,1),
                 padding='valid',
                 kernel_initializer='he_uniform',
                 activation=relu))
model.add(MaxPool2D(pool_size=(2, 2)))

# Flatten

model.add(Flatten())

# now give to the ANN

# flatten to Hidden layer - 1
model.add(Dense(units=300, activation=relu, kernel_initializer='he_uniform'))

# Hidden Layer - 1 to Hidden Layer - 2
model.add(Dense(units=200, activation=relu, kernel_initializer='he_uniform'))

# Hidden Layer - 2 to Hidden Layer - 3
model.add(Dense(units=100, activation=relu, kernel_initializer='he_uniform'))

# Hidden Layer - 3 to Hidden Layer - 4
model.add(Dense(units=50, activation=relu, kernel_initializer='he_uniform'))

# Hidden Layer - 4 to Hidden Layer - 5
model.add(Dense(units=20, activation=relu, kernel_initializer='he_uniform'))

# Hidden Layer - 5 to Hidden Layer - 6
model.add(Dense(units=len(labels), kernel_initializer='glorot_uniform', activation=softmax))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['Accuracy'])
```

```python
model.fit(final_training_data, validation_data=final_valdiation_data, epochs=50)
```

```python
# =========================
# MOUNT GOOGLE DRIVE
# =========================
from google.colab import drive
drive.mount('/content/drive')

# =========================
# IMPORT LIBRARIES
# =========================
import os
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, Dense, MaxPool2D, Flatten
from tensorflow.keras.activations import relu, softmax
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

# =========================
# DATASET PATH
# =========================
val_data_path = "/content/dataset_split_01/val"

# =========================
# GET CLASS LABELS
# =========================
labels = [
    folder for folder in os.listdir(val_data_path)
    if os.path.isdir(os.path.join(val_data_path, folder))
]

print("Classes:")
print(labels)

# =========================
# VALIDATION GENERATOR
# =========================
validation_data_preprocessing = ImageDataGenerator(rescale=1/255)

final_valdiation_data = validation_data_preprocessing.flow_from_directory(
    val_data_path,
    target_size=(200,200),
    classes=labels,
    class_mode='categorical',
    batch_size=32,
    shuffle=False
)

# =========================
# REBUILD CNN MODEL
# =========================
model = Sequential()

# Layer 1
model.add(Conv2D(
    filters=500,
    kernel_size=(3,3),
    strides=(1,1),
    padding='valid',
    kernel_initializer='he_uniform',
    activation=relu,
    input_shape=(200,200,3)
))
model.add(MaxPool2D(pool_size=(2,2)))

# Layer 2
model.add(Conv2D(
    filters=250,
    kernel_size=(3,3),
    activation=relu,
    kernel_initializer='he_uniform'
))
model.add(MaxPool2D(pool_size=(2,2)))

# Layer 3
model.add(Conv2D(
    filters=100,
    kernel_size=(3,3),
    activation=relu,
    kernel_initializer='he_uniform'
))
model.add(MaxPool2D(pool_size=(2,2)))

# Layer 4
model.add(Conv2D(
    filters=50,
    kernel_size=(3,3),
    activation=relu,
    kernel_initializer='he_uniform'
))
model.add(MaxPool2D(pool_size=(2,2)))

# Layer 5
model.add(Conv2D(
    filters=20,
    kernel_size=(3,3),
    activation=relu,
    kernel_initializer='he_uniform'
))
model.add(MaxPool2D(pool_size=(2,2)))

# Flatten
model.add(Flatten())

# Dense Layers
model.add(Dense(300, activation=relu))
model.add(Dense(200, activation=relu))
model.add(Dense(100, activation=relu))
model.add(Dense(50, activation=relu))
model.add(Dense(20, activation=relu))

# Output Layer
model.add(Dense(len(labels), activation=softmax))

# =========================
# LOAD WEIGHTS
# =========================
model.load_weights(
    "/content/food_classification_weights.weights.h5"
)

print("Weights Loaded Successfully")

# =========================
# PREDICTIONS
# =========================
y_true = final_valdiation_data.classes

y_pred_probs = model.predict(final_valdiation_data)

y_pred = np.argmax(y_pred_probs, axis=1)

# =========================
# METRICS
# =========================
cm = confusion_matrix(y_true, y_pred)

acc = accuracy_score(y_true, y_pred)

report = classification_report(
    y_true,
    y_pred,
    target_names=list(final_valdiation_data.class_indices.keys())
)

# =========================
# PRINT RESULTS
# =========================
print("\n================ CONFUSION MATRIX ================\n")
print(cm)

print("\n================ ACCURACY ================\n")
print(acc)

print("\n================ CLASSIFICATION REPORT ================\n")
print(report)
```
